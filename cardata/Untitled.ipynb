```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor as dtr
from sklearn.model_selection import train_test_split as tts
from sklearn.linear_model import LinearRegression as lr 
from sklearn.ensemble import RandomForestRegressor as rf
from sklearn.metrics import mean_squared_error, r2_score
```

```python
car= pd.read_csv("car data.csv")
```

```python
car.info()
```

```python
bike = ['royal enfield','bajaj','ktm','hyosung','yamaha','tvs','hero','apache','bullet','pulsar','jawa','activa','access125','splendor']
car['car_name']=car['Car_Name'].str.lower()
car = car[~car['car_name'].str.contains('|'.join(bike))]
```

```python
# car = car.copy()   # important!

car['Car_Name'] = car['Car_Name'].astype(str)
car['Brand'] = car['Car_Name'].str.split().str[0].str.capitalize()
```

```python
brand_goodwill = {
    "Maruti": 8,
    "Hyundai": 8,
    "Honda": 8,
    "Toyota": 9,
    "Mahindra": 7,
    "Tata": 8,
    "Ford": 7,
    "Volkswagen": 8,
    "BMW": 9,
    "Audi": 9,
    "Mercedes": 10
}

car["Brand_Goodwill"] = car["Brand"].map(brand_goodwill)
```

```python
car.info()
```

```python
hp_map = {
    "ritz": 85,
    "swift": 82,
    "wagon r": 67,
    "sx4": 104,
    "ciaz": 103
}

car["Horsepower"] = car["Car_Name"].str.lower().map(hp_map)
```

```python
car.info()
```

```python
mileage_map = {
    "swift": 21,
    "ciaz": 20,
    "wagon r": 22,
    "ritz": 18,
    "sx4": 16
}

car["Mileage"] = car["Car_Name"].str.lower().map(mileage_map)

car.info()
```

```python
y= car.Present_Price
```

```python
features = ['Brand_Goodwill','Horsepower','Mileage']
X = car[features]
X = car[features]
X = X.fillna(X.median())
```

```python
X_train,X_test,y_train,y_test = tts(X,y,random_state=1)
```

```python
model = rf(random_state=1)
```

```python
model = lr()
```

```python
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_predict)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_predict)

print("RMSE:", rmse)
print("R² Score:", r2)
```
